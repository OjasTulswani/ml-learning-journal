```python
df_country = pd.read_excel('Zomatodataset/Country-Code.xlsx')
df_country.head()
```

```python
df.groupby(['Country Code']).count().shape
```

```python
final_df = pd.merge(df, df_country, on='Country Code', how='left')
```

```python
df.shape
```

```python
df_country.shape
```

```python
final_df.columns
```

```python
country_names = final_df.Country.value_counts().index
```

```python
country_values = final_df.Country.value_counts().values
```

```python
plt.pie(country_values[:3], labels=country_names[:3], autopct='%0.2f%%')
```

##### observation: India has the highest number of transactions on Zomato followed by US and then UK

```python
final_df.columns
```

```python
ratings = final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns={0: 'Rating count'})
```

```python
ratings.head()
```

#### observation: 
- white or 0 means no rating 
- red or poor are from 1.8 to 2.4
- orange or average from 2.5 to 3.4
- yellow or good are from 3.5 to 3.9
- green or v.good are from 4.0 to 4.4 
- d.green or excllent from 4.5 to 4.9


```python
sns.barplot(data=ratings, x="Aggregate rating", y="Rating count", hue="Rating color", palette=['grey', 'red', 'orange', 'yellow', 'green', 'darkgreen'])
plt.rcParams['figure.figsize'] = (12, 10)
```

```python
### find the countris with zero ratings
final_df[final_df['Aggregate rating'] == 0].groupby('Country').size()
```

```python
### find put which currency is used by which country
final_df[['Country', 'Currency']].groupby(['Country', 'Currency']).size().reset_index()
```

```python
# which country do have online delivery option
final_df[['Country', 'Has Online delivery']].groupby(['Country', 'Has Online delivery']).size().reset_index()
```
